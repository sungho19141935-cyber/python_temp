```python
def calculate_shipping(order_amount):
    if order_amount >= 50000:
        return 0
    return 3000

    #. 경계값 테스트라는 게 왜 필요한지
print(calculate_shipping(49999))   
print(calculate_shipping(50000))  
print(calculate_shipping(50001))   
#"코드가 돌아간다"랑 "코드가 맞다"는 다른 얘기라는 것
```

```python
tests = [
    (40000, 3000),
    (49999, 3000),
    (50000, 0),
    (50001, 0),
    (60000, 0),
]
for order_amount, expected in tests:
    actual = calculate_shipping(order_amount)
    print(order_amount, actual == expected)
#즉 내가 생각했을 때는 이 디버깅은 코드가 잘 돌아가는 것처럼 보이지만 테스트를 돌려 그게 실제로 문제없이 돌아가고 있는건지를 확인하는 단계임을 알겠음
```

```python
# 클로드가 제시한 다른 유형의 디버깅

def check_membership(answer):
    if answer == "y":
        return "회원 할인 5% 적용"
    return "일반 요금 적용"

print(check_membership(input("회원이신가요? (y/n): "))) # y 외에 대문자 Y 라던지 스페이스바가 눌린 y인 비정상값이 들어갈 수 있음
```

디버깅 순서 !!!!!!!!!!

문제 함수 : check_membership()
입력      : "Y" (대문자)
실제 결과 : 일반 요금 적용
예상 결과 : 회원 할인 5% 적용

```python
tests = [
    ("y", "회원 할인 5% 적용"),
    ("Y", "회원 할인 5% 적용"),
    (" y ", "회원 할인 5% 적용"),
    ("n", "일반 요금 적용"),
    ("N", "일반 요금 적용"),
]
for answer, expected in tests:
    actual = check_membership(answer)
    print(repr(answer), actual == expected)
```
