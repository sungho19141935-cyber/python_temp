재고 부족 상황 추가하기

현재 재고: 5
판매 요청: 8

요구사항:

 

• 판매 가능한 경우에만 재고 감소
• 재고 부족이면 재고 유지
• 성공/실패를 반환값으로 구분
• 판매 후 show_info()로 상태 확인

```python
class Product:
    def __init__(self, name, price, stock):
        self.name = name
        self.price = price
        self.stock = stock

    def sell(self, quantity):
        if quantity <= self.stock:
            self.stock -= quantity
            return True
        return False

    def show_info(self):
        print(f"{self.name}:{self.price:,}원 / 재고 {self.stock}개")

product = Product("키보드", 45000, 5)
product.show_info()

print(product.sell(8))

product.show_info()

    
        
```
